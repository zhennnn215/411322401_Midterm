````python
# ============================================================
# 第 2 格：建立資料庫與初始資料
# 提醒：每次執行都要 DROP TABLE IF EXISTS 再重建
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
import sqlite3

def init_db():
    conn = sqlite3.connect("activities.db")
    cursor = conn.cursor()
    try:
        # 每次執行都重建資料表，確保資料乾淨
        cursor.execute("DROP TABLE IF EXISTS registrations")
        cursor.execute("DROP TABLE IF EXISTS activities")

        # 建立活動表
        cursor.execute("""
            CREATE TABLE activities (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                name TEXT NOT NULL,
                date TEXT NOT NULL,
                capacity INTEGER NOT NULL
            )
        """)

        # 建立報名紀錄表
        cursor.execute("""
            CREATE TABLE registrations (
                id INTEGER PRIMARY KEY AUTOINCREMENT,
                student_name TEXT NOT NULL,
                activity_name TEXT NOT NULL,
                count INTEGER NOT NULL,
                status TEXT NOT NULL
            )
        """)

        # 放入初始資料
        initial_activities = [
            ('迎新晚會', '2026-11-20', 6),
            ('程式工作坊', '2026-11-25', 4),
            ('籃球友誼賽', '2026-11-28', 3)
        ]
        cursor.executemany("INSERT INTO activities (name, date, capacity) VALUES (?, ?, ?)", initial_activities)

        conn.commit()
        print("資料庫與初始資料建立成功！")
    finally:
        conn.close()

# 執行初始化
init_db()
````

````python
# ============================================================
# 第 3 格：顯示資料表
# 提醒：定義 show_tables() 並呼叫一次
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (3) 顯示資料表：使用 pandas 讀取並印出目前的活動表與報名紀錄表內容
import sqlite3
import pandas as pd

def show_tables():
    conn = sqlite3.connect("activities.db")
    try:
        print("=== 【活動表】 ===")
        df_activities = pd.read_sql_query("SELECT * FROM activities", conn)
        display(df_activities)

        print("\n=== 【報名紀錄表】 ===")
        df_registrations = pd.read_sql_query("SELECT * FROM registrations", conn)
        display(df_registrations)
    finally:
        conn.close()

# 執行顯示
show_tables()
````

````python
# ============================================================
# 第 4 格：AI 解讀函式
# 提醒：回傳 (動作, 活動, 人數, 原始回覆)；不可直接相信 AI 的回覆
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (4) AI 解讀函式：透過 Groq AI 解析使用者的一句話，並使用快取避免重複呼叫、429 重試機制
import re
import time

# 建立快取字典，避免同一句話重複呼叫 AI
ai_cache = {}

def parse_user_intent(text):
    # 若該句話已經查詢過，直接從快取回傳
    if text in ai_cache:
        return ai_cache[text]

    prompt = f"""
    請從以下學生活動的一句話中萃取出三個資訊：動作、活動、人數。
    1. 動作只能是：報名、取消、查詢、統計、未知
    2. 活動只能是：迎新晚會、程式工作坊、籃球友誼賽、無
    3. 人數請填入整數（若提到我和兩個朋友代表3人；若沒提到人數，預設為1）。

    請嚴格依照以下格式回傳，不要有其他廢話：
    動作,活動,人數

    使用者說的話：「{text}」
    """

    response_text = ""
    # 最多重試 3 次 (針對 429 過量錯誤)
    for attempt in range(3):
        try:
            completion = client.chat.completions.create(
                model=MODEL,
                messages=[{"role": "user", "content": prompt}],
                temperature=0,
                reasoning_effort=REASONING_EFFORT
            )
            response_text = completion.choices[0].message.content.strip()
            time.sleep(2)  # 每次呼叫後暫停 2 秒
            break
        except Exception as e:
            if "429" in str(e) or "rate_limit" in str(e).lower():
                print(f"遇到 API 額度限制 (429)，等待 30 秒後重試 (第 {attempt+1} 次)...")
                time.sleep(30)
            else:
                print(f"AI 呼叫發生錯誤: {e}")
                break

    # 安全防範：解析 AI 回傳的字串
    action, activity, count = "未知", "無", 1
    parts = response_text.split(",")
    if len(parts) >= 3:
        raw_action = parts[0].strip()
        raw_activity = parts[1].strip()
        raw_count_str = parts[2].strip()

        # 檢查動作允許清單
        if raw_action in ["報名", "取消", "查詢", "統計"]:
            action = raw_action

        # 檢查活動允許清單
        if raw_activity in ["迎新晚會", "程式工作坊", "籃球友誼賽"]:
            activity = raw_activity

        # 用正規表示式抓出數字作為人數
        match_num = re.search(r'\d+', raw_count_str)
        if match_num:
            count = int(match_num.group())

    result = (action, activity, count)
    ai_cache[text] = result  # 存入快取
    return response_text, result
````

````python
# ============================================================
# 第 5 格：AI 解讀測試
# 提醒：8～10 句，要與規格書 7-1 一致
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (5) AI 解讀測試：執行 8~10 句測試，印出原始回覆、解讀結果與通過統計
test_sentences = [
    ("我和兩個朋友想報名迎新晚會", "報名", "迎新晚會", 3),
    ("幫我報名程式工作坊", "報名", "程式工作坊", 1),
    ("取消迎新晚會", "取消", "迎新晚會", 1),
    ("查詢籃球友誼賽剩餘名額", "查詢", "籃球友誼賽", 1),
    ("各活動報名狀況", "統計", "無", 1),
    ("隨便亂打測試字串", "未知", "無", 1),
    ("我要報名迎新晚會五個人", "報名", "迎新晚會", 5),
    ("請問程式工作坊還有名額嗎", "查詢", "程式工作坊", 1),
    ("幫我取消程式工作坊", "取消", "程式工作坊", 1),
    ("我要報名籃球友誼賽我和一個朋友", "報名", "籃球友誼賽", 2)
]

passed_count = 0
print("=== 【AI 解讀測試開始】 ===\n")

for i, (text, exp_action, exp_activity, exp_count) in enumerate(test_sentences, 1):
    raw_res, (act, actv, cnt) = parse_user_intent(text)

    # 檢查是否符合預期（支援 "*" 表示不在意該項）
    match_action = (exp_action == "*" or act == exp_action)
    match_activity = (exp_activity == "*" or actv == exp_activity)
    match_count = (exp_count == "*" or cnt == exp_count)

    is_passed = match_action and match_activity and match_count
    if is_passed:
        passed_count += 1
        status_str = "【通過】"
    else:
        status_str = "【未通過】"

    print(f"測試 {i}: {status_str}")
    print(f"  使用者說: {text}")
    print(f"  AI 原始回覆: {raw_res}")
    print(f"  系統解讀: 動作={act}, 活動={actv}, 人數={cnt}")
    print(f"  預期結果: 動作={exp_action}, 活動={exp_activity}, 人數={exp_count}\n")

print(f"=== 測試結果摘要：總共 10 句，成功通過 {passed_count} 句 ===")
````

````python
# ============================================================
# 第 6 格：功能函式
# 提醒：fetchone() 後先檢查 None；用 try/finally 關閉連線；統計用 JOIN＋GROUP BY
# ============================================================
# （把 AI 產生、你看過並理解的程式貼在這裡）
# (6) 功能函式：實作報名、取消、查詢與統計的資料庫商業邏輯與規則檢查
import sqlite3

def process_system_request(student_name, user_input):
    # 1. 透過 AI 解析意圖
    _, (action, activity_name, count) = parse_user_intent(user_input)

    # 如果使用者沒輸入姓名（且非統計查詢），請其補上
    if not student_name.strip() and action != "統計":
        return "請先輸入您的姓名才能進行操作！"

    # 如果無法判斷或動作為未知，回覆目前有哪些活動
    if action == "未知" or activity_name == "無" and action != "統計":
        return "無法辨識您的需求。目前有的活動為：迎新晚會、程式工作坊、籃球友誼賽。"

    conn = sqlite3.connect("activities.db")
    try:
        cursor = conn.cursor()

        # -------------------- 統計查詢 (進階 A3) --------------------
        if action == "統計":
            cursor.execute("""
                SELECT
                    a.name,
                    a.date,
                    a.capacity,
                    COALESCE(SUM(CASE WHEN r.status = '已報名' THEN r.count ELSE 0 END), 0) as booked_count
                FROM activities a
                LEFT JOIN registrations r ON a.name = r.activity_name
                GROUP BY a.id, a.name, a.date, a.capacity
            """)
            rows = cursor.fetchall()
            result_str = "=== 各活動報名狀況統計 ===\n"
            for row in rows:
                name, date, cap, booked = row[0], row[1], row[2], row[3]
                remaining = cap - booked
                result_str += f"- 【{name}】 日期: {date} | 總名額: {cap} | 已報名人數: {booked} | 剩餘名額: {remaining}\n"
            return result_str.strip()

        # -------------------- 查詢單一活動剩餘名額 --------------------
        if action == "查詢":
            cursor.execute("SELECT date, capacity FROM activities WHERE name = ?", (activity_name,))
            act_row = cursor.fetchone()
            if act_row is None:
                return f"找不到活動：{activity_name}"
            date, capacity = act_row[0], act_row[1]

            cursor.execute("""
                SELECT COALESCE(SUM(count), 0) FROM registrations
                WHERE activity_name = ? AND status = '已報名'
            """, (activity_name,))
            sum_row = cursor.fetchone()
            booked = sum_row[0] if sum_row else 0

            remaining = capacity - booked
            return f"【{activity_name}】於 {date} 舉辦，總名額 {capacity} 人，目前剩餘名額：{remaining} 人。"

        # -------------------- 報名功能 (含進階 A1, A2 規則) --------------------
        if action == "報名":
            # 檢查人數欄位規則：不在 1~3 之間由程式拒選 (A1)
            if count < 1 or count > 3:
                return f"報名失敗：人數必須介於 1 到 3 之間（您輸入了 {count} 人）。"

            # 檢查活動是否存在並取得總容量
            cursor.execute("SELECT capacity FROM activities WHERE name = ?", (activity_name,))
            act_row = cursor.fetchone()
            if act_row is None:
                return f"報名失敗：找不到活動「{activity_name}」。"
            capacity = act_row[0]

            # 檢查該學生是否已經「已報名」過該活動 (不可重複報名)
            cursor.execute("""
                SELECT id FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity_name))
            if cursor.fetchone() is not None:
                return f"報名失敗：{student_name} 已經報名過【{activity_name}】了，不可重複報名。"

            # 檢查每人最多同時報名 2 個活動（已取消的不計）(A2)
            cursor.execute("""
                SELECT COUNT(*) FROM registrations
                WHERE student_name = ? AND status = '已報名'
            """, (student_name,))
            user_active_count = cursor.fetchone()[0]
            if user_active_count >= 2:
                return f"報名失敗：{student_name} 目前已同時報名了 {user_active_count} 個活動（每人最多同時報名 2 個）。"

            # 檢查名額是否足夠 (以人數加總計算)
            cursor.execute("""
                SELECT COALESCE(SUM(count), 0) FROM registrations
                WHERE activity_name = ? AND status = '已報名'
            """, (activity_name,))
            sum_row = cursor.fetchone()
            current_booked = sum_row[0] if sum_row else 0

            if current_booked + count > capacity:
                remaining = capacity - current_booked
                return f"報名失敗：【{activity_name}】名額不足！目前剩餘名額僅剩 {remaining} 人（您欲報名 {count} 人）。"

            # 通過所有檢查，寫入資料庫
            cursor.execute("""
                INSERT INTO registrations (student_name, activity_name, count, status)
                VALUES (?, ?, ?, '已報名')
            """, (student_name, activity_name, count))
            conn.commit()
            return f"報名成功！{student_name} 已成功報名【{activity_name}】，共 {count} 人。"

        # -------------------- 取消功能 --------------------
        if action == "取消":
            # 檢查該學生是否有該活動的「已報名」紀錄
            cursor.execute("""
                SELECT id FROM registrations
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity_name))
            reg_row = cursor.fetchone()
            if reg_row is None:
                return f"取消失敗：查無 {student_name} 報名【{activity_name}】的有效紀錄。"

            # 更新狀態為「已取消」（釋出名額）
            cursor.execute("""
                UPDATE registrations SET status = '已取消'
                WHERE student_name = ? AND activity_name = ? AND status = '已報名'
            """, (student_name, activity_name))
            conn.commit()
            return f"取消成功！{student_name} 已取消【{activity_name}】的報名，名額已釋出。"

    finally:
        conn.close()

    return "系統發生未預期的錯誤。"
````
